```python


import re
import gradio as gr

URGENCY_KEYWORDS = [
    "urgent", "immediately", "act now", "verify now",
    "account suspended", "limited time",
    "জরুরি", "এখনই", "অবিলম্বে", "যাচাই করুন"
]

SENSITIVE_KEYWORDS = [
    "otp", "password", "pin", "bank account",
    "recovery code", "secret key",
    "বিকাশ পিন", "নগদ পিন", "ওটিপি", "পাসওয়ার্ড"
]

MONEY_KEYWORDS = [
    "send money", "transfer money", "claim your prize",
    "payment required", "processing fee",
    "টাকা পাঠান", "পুরস্কার জিতেছেন", "টাকা দিন"
]

def analyze_message(message):
    if not message or not message.strip():
        return "No Input", "Please enter a message to analyze."

    text = message.lower()
    score = 0
    reasons = []

    # Detect URLs
    urls = re.findall(r"(?:https?://|www\.)\S+", text)
    if urls:
        score += 2
        reasons.append(
            f"Contains {len(urls)} URL(s). Check the destination carefully."
        )

    # Detect urgency and pressure tactics
    found_urgency = [
        word for word in URGENCY_KEYWORDS if word in text
    ]
    if found_urgency:
        score += 1
        reasons.append(
            "Uses urgency or pressure language: "
            + ", ".join(found_urgency)
        )

    # Detect requests involving sensitive information
    found_sensitive = [
        word for word in SENSITIVE_KEYWORDS if word in text
    ]
    if found_sensitive:
        score += 2
        reasons.append(
            "Mentions sensitive information or security credentials: "
            + ", ".join(found_sensitive)
        )

    # Detect financial requests or suspicious prize claims
    found_money = [
        word for word in MONEY_KEYWORDS if word in text
    ]
    if found_money:
        score += 2
        reasons.append(
            "Contains financial or prize-related language: "
            + ", ".join(found_money)
        )

    # Assign a preliminary risk level
    if score >= 4:
        risk = "HIGH RISK"
        advice = (
            "Do not click suspicious links or share passwords, "
            "PINs, or one-time codes. Verify the message through "
            "the organization's official website or phone number."
        )
    elif score >= 2:
        risk = "MEDIUM RISK"
        advice = (
            "Verify the sender and destination independently. "
            "Avoid sharing sensitive information until verified."
        )
    else:
        risk = "LOW SIGNAL"
        advice = (
            "No strong indicators were detected by these basic rules. "
            "This does not prove the message is safe."
        )

    if not reasons:
        reasons.append(
            "No indicators covered by the current rules were detected."
        )

    report = (
        f"Risk Assessment: {risk}\n\n"
        "Indicators Detected:\n"
        + "\n".join(f"- {reason}" for reason in reasons)
        + "\n\nRecommended Action:\n"
        + advice
        + "\n\nLimitations:\n"
        "This is a rule-based prototype, not an AI-trained classifier. "
        "It may miss scams or flag legitimate messages. "
        "Its risk score is a heuristic, not a verified probability."
    )

    return risk, report


demo = gr.Interface(
    fn=analyze_message,
    inputs=gr.Textbox(
        lines=5,
        label="Message to Analyze",
        placeholder="Paste a suspicious SMS, email, or message here..."
    ),
    outputs=[
        gr.Textbox(label="Risk Level"),
        gr.Textbox(label="Analysis Report", lines=12)
    ],
    title="PhishShield AI",
    description=(
        "A bilingual scam and phishing risk checker "
        "for English and Bangla messages."
    ),
    examples=[
        [
            "Your bank account is suspended! Verify now at "
            "http://example.com and share your OTP immediately."
        ],
        [
            "আপনার অ্যাকাউন্ট বন্ধ হবে। এখনই ওটিপি ও পিন দিন।"
        ],
        [
            "Hello, your appointment is confirmed for tomorrow at 10 AM."
        ]
    ]
)

demo.launch(share=True)
```

```python


import requests
from urllib.parse import quote

EUROPE_PMC_URL = (
    "https://www.ebi.ac.uk/europepmc/webservices/rest/search"
)

def search_oncology_papers(query, limit=5):
    """Search Europe PMC for oncology research papers."""
    query = (query or "").strip()

    if not query:
        return []

    if len(query) > 300:
        raise ValueError("Search query must be under 300 characters.")

    response = requests.get(
        EUROPE_PMC_URL,
        params={
            "query": query,
            "format": "json",
            "resultType": "core",
            "pageSize": max(1, min(int(limit), 10)),
            "sort_date": "y",
        },
        timeout=30,
    )
    response.raise_for_status()

    records = response.json().get("resultList", {}).get("result", [])
    papers = []

    for item in records:
        paper_id = item.get("id", "")
        source = item.get("source", "")
        title = item.get("title", "Untitled")
        abstract = item.get("abstractText", "")
        doi = item.get("doi", "")
        year = item.get("firstPublicationDate", "")[:4]

        if source == "MED":
            url = f"https://europepmc.org/article/MED/{paper_id}"
        elif source == "PMC":
            url = f"https://europepmc.org/articles/{paper_id}"
        elif doi:
            url = f"https://doi.org/{doi}"
        else:
            url = (
                "https://europepmc.org/search?query="
                + quote(title)
            )

        papers.append({
            "title": title,
            "year": year,
            "doi": doi,
            "abstract": abstract,
            "url": url,
        })

    return papers


def online_research_answer(question, limit=5):
    papers = search_oncology_papers(question, limit=limit)

    if not papers:
        return (
            "No matching papers were returned. Try a broader query. "
            "This does not prove that no relevant research exists."
        )

    evidence_blocks = []

    for i, paper in enumerate(papers, start=1):
        evidence_blocks.append(
            f"""SOURCE {i}
Title: {paper['title']}
Year: {paper['year'] or 'Not available'}
DOI: {paper['doi'] or 'Not available'}
URL: {paper['url']}
Abstract:
{paper['abstract'] or 'No abstract available'}
"""
        )

    evidence = "\n\n".join(evidence_blocks)

    prompt = f"""
Analyze the following retrieved scientific records in English.

Research question:
{question}

Retrieved records:
{evidence}

Requirements:
- Use only the supplied records for claims about these specific papers.
- Cite evidence using [1], [2], etc., matching the source numbers.
- Distinguish paper findings from your own interpretation.
- State when abstracts are missing or evidence is insufficient.
- Do not invent methods, results, citations, or clinical recommendations.
- Identify relevant study limitations only when supported by the record.
- This is research support, not medical advice.
"""

    answer = analyze_research(
        question=prompt,
        source_text=evidence,
        max_tokens=1000,
    )

    references = "\n\n## Retrieved source links\n"
    for i, paper in enumerate(papers, start=1):
        references += (
            f"\n{i}. [{paper['title']}]({paper['url']})"
            f" — {paper['year'] or 'Year unavailable'}"
            f"{'; DOI: ' + paper['doi'] if paper['doi'] else ''}\n"
        )

    return answer + references
```

```python


import requests

API_URL = "https://api.featherless.ai/v1/chat/completions"
MODEL = "Qwen/Qwen3.5-4B"

SYSTEM_PROMPT = """
You are Oncology Research AI, an English-language scientific research
assistant specializing in oncology literature analysis.

Use retrieved scientific evidence carefully.
Do not invent facts, citations, statistics, or paper findings.
Distinguish source-supported findings from interpretation.
Explain uncertainty and limitations.
Do not diagnose patients or recommend personalized cancer treatment,
medications, or chemotherapy doses.
This tool is a research prototype, not a clinically validated system.
"""

def analyze_research(question, source_text, max_tokens=700):
    if not FEATHERLESS_API_KEY:
        raise ValueError("FEATHERLESS_API_KEY is missing.")

    payload = {
        "model": MODEL,
        "messages": [
            {"role": "system", "content": SYSTEM_PROMPT},
            {
                "role": "user",
                "content": (
                    f"Research question:\n{question}\n\n"
                    f"Retrieved scientific evidence:\n{source_text}\n\n"
                    "Analyze the evidence in English. Cite sources using "
                    "their supplied numbers. State uncertainty and limitations. "
                    "Never invent citations or claim to have verified information "
                    "that is not present in the supplied evidence."
                ),
            },
        ],
        "max_tokens": max_tokens,
        "temperature": 0.1,
    }

    response = requests.post(
        API_URL,
        headers={
            "Authorization": f"Bearer {FEATHERLESS_API_KEY}",
            "Content-Type": "application/json",
        },
        json=payload,
        timeout=120,
    )

    if not response.ok:
        raise RuntimeError(
            f"Featherless API error: HTTP {response.status_code}. "
            "Check API access, model availability, and account limits."
        )

    data = response.json()
    choices = data.get("choices", [])

    if not choices:
        raise RuntimeError("The API returned no completion.")

    content = choices[0].get("message", {}).get("content", "")
    if not content:
        raise RuntimeError("The model returned an empty answer.")

    return content.strip()

print("analyze_research() is ready.")
```

```python


import requests

API_URL = "https://api.featherless.ai/v1/chat/completions"
MODEL = "Qwen/Qwen3.5-4B"

SYSTEM_PROMPT = """
You are Oncology Research AI, an English-language assistant
for scientific oncology research and literature analysis.
Use only evidence supplied in the prompt for claims about
specific papers. Never invent citations or research findings.
Clearly state uncertainty and limitations.
Do not diagnose patients or prescribe personalized treatment.
This prototype is not clinically validated.
"""

def analyze_research(question, source_text, max_tokens=900):
    if not FEATHERLESS_API_KEY:
        raise ValueError("API key is missing. Load it from Colab Secrets.")

    response = requests.post(
        API_URL,
        headers={
            "Authorization": f"Bearer {FEATHERLESS_API_KEY}",
            "Content-Type": "application/json",
        },
        json={
            "model": MODEL,
            "messages": [
                {"role": "system", "content": SYSTEM_PROMPT},
                {
                    "role": "user",
                    "content": (
                        f"Research question:\n{question}\n\n"
                        f"Evidence:\n{source_text}\n\n"
                        "Write a concise final answer in English. "
                        "Distinguish evidence from interpretation. "
                        "Do not invent citations. Do not include reasoning."
                    ),
                },
            ],
            "max_tokens": max_tokens,
            "temperature": 0.1,
        },
        timeout=120,
    )

    if not response.ok:
        raise RuntimeError(
            f"API error HTTP {response.status_code}: {response.text[:500]}"
        )

    data = response.json()
    choices = data.get("choices", [])

    if not choices:
        raise RuntimeError("The API returned no choices.")

    message = choices[0].get("message") or {}
    answer = message.get("content", "")

    if isinstance(answer, list):
        answer = "\n".join(
            part.get("text", "")
            for part in answer
            if isinstance(part, dict)
        )

    if not isinstance(answer, str) or not answer.strip():
        raise RuntimeError(
            "Final answer is empty. Finish reason: "
            + str(choices[0].get("finish_reason"))
        )

    return answer.strip()

print("analyze_research() is ready.")
```

```python

import requests

def search_oncology_papers(query, limit=3):
    endpoint = "https://www.ebi.ac.uk/europepmc/webservices/rest/search"
    params = {
        "query": query,
        "format": "json",
        "resultType": "core",
        "pageSize": min(max(int(limit), 1), 10)
    }

    response = requests.get(endpoint, params=params, timeout=30)
    response.raise_for_status()
    results = response.json().get("resultList", {}).get("result", [])

    papers = []
    for p in results:
        doi = p.get("doi", "")
        paper_id = p.get("id", "")
        source = p.get("source", "MED")
        link = (
            "https://doi.org/" + doi
            if doi else
            "https://europepmc.org/article/" + source + "/" + paper_id
        )
        papers.append({
            "title": p.get("title", "Untitled"),
            "year": p.get("firstPublicationDate", "Unknown"),
            "abstract": p.get("abstractText", "No abstract available."),
            "url": link
        })
    return papers


def online_research_answer(question, limit=3):
    papers = search_oncology_papers(question, limit)
    if not papers:
        return "No papers found. Try a broader search query."

    evidence = "\n\n".join(
        f"Title: {p['title']}\nYear: {p['year']}\n"
        f"Abstract: {p['abstract'][:1500]}"
        for p in papers
    )

    summary = analyze_research(question, evidence, max_tokens=500)
    refs = "\n\n".join(
        f"{i}. {p['title']} ({p['year']})\n{p['url']}"
        for i, p in enumerate(papers, 1)
    )
    return (
        "## Research Summary\n\n" + summary
        + "\n\n## Retrieved Papers\n\n" + refs
        + "\n\nDisclaimer: Research prototype, not a clinical decision tool."
    )

print("Functions repaired successfully.")
```
